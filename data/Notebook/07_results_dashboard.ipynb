```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
landing_sites = pd.read_csv("final_landing_sites.csv")

landing_sites.head()
```

```python
top10 = landing_sites.head(10)

top10
```

```python
plt.figure(figsize=(12,5))

plt.bar(
    range(len(top10)),
    top10["Final_Score"]
)

plt.title("Top 10 Landing Sites")
plt.xlabel("Site Index")
plt.ylabel("Final Score")

plt.grid(True)

plt.show()
```

```python
plt.figure(figsize=(12,5))

plt.hist(
    landing_sites["Ice_Probability"],
    bins=30
)

plt.title("Ice Probability Distribution")
plt.xlabel("Ice Probability")
plt.ylabel("Count")

plt.grid(True)

plt.show()
```

```python
top_ice = landing_sites.sort_values(
    by="Ice_Probability",
    ascending=False
)

top_ice.head(20)
```

```python
landing_sites[
    [
        "Signal",
        "Ice_Probability",
        "Final_Score"
    ]
].describe()
```

```python
print("Total Samples:", len(landing_sites))
print("Top Landing Sites:", len(top10))
print("Highest Ice Probability:",
      landing_sites["Ice_Probability"].max())
print("Highest Final Score:",
      landing_sites["Final_Score"].max())
```
