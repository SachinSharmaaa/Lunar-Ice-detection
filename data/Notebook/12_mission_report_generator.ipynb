```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
landing_sites = pd.read_csv("final_landing_sites.csv")
ice_map = pd.read_csv("lunar_ice_probability_map.csv")

print("Landing Sites:", len(landing_sites))
print("Top Ice Sites:", len(ice_map))
```

```python
best_site = ice_map.iloc[0]

print("=== BEST LANDING SITE ===")
print(best_site)
```

```python
plt.figure(figsize=(10,5))

plt.bar(
    range(len(ice_map)),
    ice_map["Ice_Probability"]
)

plt.title("Top Lunar Ice Candidate Sites")
plt.xlabel("Site Rank")
plt.ylabel("Ice Probability")

plt.show()
```

```python
summary = pd.DataFrame({
    "Metric":[
        "Total Landing Sites",
        "Top Candidate Sites",
        "Radar Data Used",
        "ML Classification"
    ],
    "Value":[
        len(landing_sites),
        len(ice_map),
        "DFSAR PRD",
        "Completed"
    ]
})

summary
```
