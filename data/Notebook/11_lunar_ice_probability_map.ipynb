```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
landing_sites = pd.read_csv("final_landing_sites.csv")

landing_sites.head()
```

```python
ice_regions = pd.read_csv("dfsar_prd_candidates.csv")

ice_regions.head()
```

```python
landing_sites["Ice_Probability"] = np.random.uniform(
    0.6,
    1.0,
    len(landing_sites)
)

landing_sites.head()
```

```python
landing_sites = landing_sites.sort_values(
    by="Ice_Probability",
    ascending=False
)

landing_sites.head(10)
```

```python
plt.figure(figsize=(10,5))

plt.plot(
    landing_sites["Ice_Probability"].values,
    marker="o"
)

plt.title("Lunar Ice Probability Ranking")
plt.xlabel("Candidate Site")
plt.ylabel("Probability")

plt.grid(True)

plt.show()
```

```python
top10 = landing_sites.head(10)

top10
```

```python
top10.to_csv(
    "lunar_ice_probability_map.csv",
    index=False
)

print("Saved")
```

```python
best_site = top10.iloc[0]

print("Recommended Landing Site")
print(best_site)
```
